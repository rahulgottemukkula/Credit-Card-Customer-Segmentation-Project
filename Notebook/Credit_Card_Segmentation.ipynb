```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_features = scaler.fit_transform(df_copy)
df_scaled = pd.DataFrame(scaled_features, columns=df_copy.columns, index=df_copy.index)
df_scaled.head()
```

```python
from sklearn.cluster import KMeans
# Elbow Method
wcss = []

for i in range(1, 11):
    model = KMeans(
        n_clusters=i,
        init='k-means++',
        random_state=42,
        n_init=10
    )
    model.fit(df_scaled)
    wcss.append(model.inertia_)

plt.figure(figsize=(8,5))
plt.plot(range(1,11), wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.grid(True)
plt.show()
```

```python
kmeans = KMeans(
    n_clusters=4,
    init='k-means++',
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(df_scaled)

df_scaled['Cluster'] = clusters
df_copy['Cluster'] = clusters

print(df_scaled.head())
print(df_copy['Cluster'].value_counts())
```

```python
from sklearn.metrics import (silhouette_score, davies_bouldin_score, calinski_harabasz_score)
```

```python
X = df_scaled.drop(columns=['Cluster'])

silhouette_kmeans = silhouette_score(X, clusters)
davies_kmeans = davies_bouldin_score(X, clusters)
calinski_kmeans = calinski_harabasz_score(X, clusters)

print(f"Silhouette Score      : {silhouette_kmeans:.4f}")
print(f"Davies-Bouldin Index  : {davies_kmeans:.4f}")
print(f"Calinski-Harabasz Index : {calinski_kmeans:.4f}")
```

```python
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
plt.figure(figsize=(15,7))
dendrogram(linkage(df_scaled, method='ward'))
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Customers")
plt.ylabel("Euclidean Distance")
plt.show()
```

```python
# Model
hc = AgglomerativeClustering(
    n_clusters=4,
    metric='euclidean',
    linkage='ward'
)
hc_labels = hc.fit_predict(df_scaled)
# Add labels
df_copy['HC_Cluster'] = hc_labels
# Cluster counts
print(df_copy['HC_Cluster'].value_counts().sort_index())
# Cluster summary
print(df_copy.groupby('HC_Cluster').mean())
```

```python
# Evaluation Metrics
silhouette_hc = silhouette_score(df_scaled, hc_labels)
davies_hc = davies_bouldin_score(df_scaled, hc_labels)
calinski_hc = calinski_harabasz_score(df_scaled, hc_labels)

print("\nHierarchical Clustering Metrics")
print("-"*40)
print(f"Silhouette Score        : {silhouette_hc:.4f}")
print(f"Davies-Bouldin Index    : {davies_hc:.4f}")
print(f"Calinski-Harabasz Score : {calinski_hc:.4f}")
```

```python
from sklearn.cluster import DBSCAN
dbscan = DBSCAN(
    eps=1.5,
    min_samples=5
)
db_labels = dbscan.fit_predict(df_scaled)
# Add Cluster Labels
df_copy['DBSCAN_Cluster'] = db_labels
# Cluster Counts
print(df_copy['DBSCAN_Cluster'].value_counts().sort_index())
# Noise Count
noise_points = (db_labels == -1).sum()
print("Noise Points:", noise_points)
# Cluster Summary
print(
    df_copy[df_copy['DBSCAN_Cluster'] != -1]
    .groupby('DBSCAN_Cluster')
    .mean()
)
```

```python
mask = db_labels != -1

X_db = df_scaled[mask]
labels_db = db_labels[mask]

if len(set(labels_db)) > 1:
    silhouette_db = silhouette_score(X_db, labels_db)
    davies_db = davies_bouldin_score(X_db, labels_db)
    calinski_db = calinski_harabasz_score(X_db, labels_db)

    print("\nDBSCAN Evaluation Metrics")
    print("-"*40)
    print(f"Silhouette Score        : {silhouette_db:.4f}")
    print(f"Davies-Bouldin Index    : {davies_db:.4f}")
    print(f"Calinski-Harabasz Score : {calinski_db:.4f}")
else:
    print("Not enough clusters for evaluation.")
```

```python
from sklearn.decomposition import PCA
# Fit PCA
pca = PCA()
pca.fit(df_scaled)
# Explained Variance
explained_variance = pca.explained_variance_ratio_
# Cumulative Variance
cumulative_variance = explained_variance.cumsum()
# Plot
plt.figure(figsize=(10,5))
plt.plot(
    range(1, len(cumulative_variance)+1),
    cumulative_variance,
    marker='o'
)
plt.xlabel("Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA Explained Variance")
plt.grid(True)
plt.show()
```

```python
# Reduce to 2 Components
pca = PCA(n_components=2)
pca_data = pca.fit_transform(df_scaled)
pca_variance = pca.explained_variance_ratio_.sum()
df_pca = pd.DataFrame(
    pca_data,
    columns=['PC1', 'PC2']
)
print(df_pca.head())
print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)
print("\nTotal Explained Variance:")
print(pca.explained_variance_ratio_.sum())
```

```python
# PCA Scatter Plot (K-Means)
plt.figure(figsize=(8,6))
plt.scatter(
    df_pca['PC1'],
    df_pca['PC2'],
    c=clusters,
    cmap='viridis'
)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA Visualization of K-Means Clusters")
plt.colorbar(label="Cluster")
plt.show()
```
